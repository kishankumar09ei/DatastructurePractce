```python
from typing import List
def longestCommonPrefix(strs: List[str]) -> str:
    if len(strs)<=1:
        return ''
    smallest_str= min(strs,key=len)
    prefix=''
    for i in range(len(smallest_str)):
        chat_at_i=[]
        for j in strs:
            chat_at_i.append(j[i])
        #print(chat_at_i)
        if len(set(chat_at_i))==1:
            prefix+=chat_at_i[0]
            print(prefix)
        else:
            break
    return prefix

longestCommonPrefix(["flower","flow","flight"])
```

```python
## 11. Container With Most Water

from typing import List
class Solution:
    def maxArea(self, height: List[int]) -> int:
        left=0
        right=len(height)-1
        max_area=0
        while left<right:
            max_area=max(max_area, min(height[left],height[right])*(right-left))
            if height[right]>=height[left]:
                left+=1
            else:
                right-=1
        return max_area
```

```python
from typing import List
class Solution:
    def longestSubarray(self, nums: List[int]) -> int:
        l=0
        max_len=0
        num_zeros=0
        for r in range(len(nums)):
            if nums[r]==0:
                num_zeros+=1
            while num_zeros>1:                
                if nums[l]==1:
                    num_zeros-=1
                l+=1
            max_len=max(max_len,r-l)
        return max_len
```

```python
from typing import List

class Solution:
    def asteroidCollision(self, asteroids: List[int]) -> List[int]:
        stack=[]
        for i in asteroids:
            if i >0:
                stack.append(i)
            else:
                while stack and stack[-1]>0 and  abs(stack[-1])<abs(i):
                    stack.pop()
                if stack and abs(stack[-1])==abs(i):
                    stack.pop()
                elif not stack or  stack[-1]<0:
                    stack.append(i)
        return stack

```

```python
class Solution:
    def backspaceCompare(self, s: str, t: str) -> bool:
        stack1,stack2=[],[]
        for i in s:
            if stack1 and i=='#':
                stack1.pop()
            else:
                stack1.append(i)

        for i in t:
            if stack2 and i=='#':
                stack2.pop()
            else:
                stack2.append(i)
        print(f"{"".join(stack1)},{"".join(stack2)}")
        return stack1==stack2
```
